```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["HF_TOKEN"] = os.getenv("HF_TOKEN")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text="Subhash Chandra Bose, popularly known as Netaji, was one of the most prominent and revolutionary leaders of India’s freedom struggle. He was born on 23 January 1897 in Cuttack, Odisha, into a well-educated and respected Bengali family. From a young age, Bose was deeply influenced by the ideas of patriotism, sacrifice, discipline, and service to the nation. He studied at Presidency College in Calcutta and later went to England, where he prepared for and successfully passed the Indian Civil Services examination, achieving a very high rank. However, Bose did not want to serve the British government while India remained under colonial rule, so he resigned from the Indian Civil Service and returned to India to participate actively in the independence movement. He became associated with the Indian National Congress and emerged as a powerful young leader who believed that India needed a more forceful and determined struggle for complete independence. Bose was elected President of the Indian National Congress in 1938 and again in 1939, but differences arose between him and the Congress leadership, particularly over the pace and method of the freedom struggle. He believed that India should take advantage of Britain’s difficulties during the Second World War and launch a decisive struggle for independence, whereas many other leaders preferred a different approach. After his resignation from the Congress presidency, Bose continued his efforts to organize resistance against British rule. He was placed under house arrest by the British, but in 1941 he made a dramatic escape from India and eventually reached Germany. Later, he traveled to Southeast Asia and took leadership of the Indian National Army, or INA, also known as the Azad Hind Fauj, which was formed to fight British rule and liberate India. Bose established the Provisional Government of Free India, called the Azad Hind Government, and sought international support for the Indian independence movement. His famous slogan, “Give me blood, and I will give you freedom,” reflected his belief in courage, sacrifice, and total commitment to the cause of independence. He also popularized the slogan “Jai Hind,” which later became an important patriotic greeting in India. Under his leadership, the INA advanced toward India with the aim of challenging British power, although military setbacks and the changing circumstances of the Second World War prevented it from achieving its ultimate objective. Despite its military defeat, the INA and Bose had a significant psychological and political impact on India’s freedom movement, because they demonstrated an organized armed challenge to British authority and inspired many Indians. Bose remains a complex and highly debated historical figure because of his willingness to seek support from Germany, Japan, and other Axis powers during the Second World War in the belief that their assistance could help India become independent. His death remains surrounded by controversy and historical debate; he is widely believed to have died following an air crash in August 1945, but questions and alternative theories about his fate persisted for decades. Regardless of the continuing debate surrounding his final days, Subhash Chandra Bose occupies a unique place in Indian history because of his uncompromising nationalism, organizational abilities, military strategy, determination, and willingness to make great personal sacrifices for independence. His life continues to inspire people through his emphasis on courage, discipline, patriotism, and the idea that freedom requires extraordinary commitment and sacrifice."
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=200)
split_docs=text_splitter.split_text(text)
split_docs
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
from langchain_community.vectorstores import Chroma
db = Chroma.from_texts(split_docs,embeddings)
```

```python
semantic_ret = db.as_retriever()
```

```python
from langchain_community.retrievers import BM25Retriever
sparse_ret = BM25Retriever.from_texts(split_docs)
```

```python
from langchain_core.tools import retriever
from langchain_classic.retrievers import EnsembleRetriever
final_ret = EnsembleRetriever(
    retrievers=[semantic_ret,sparse_ret],
    weights=[0.7,0.3]
)
```

```python
res = final_ret.invoke("Who is subhash chandra bose?")
res
```
